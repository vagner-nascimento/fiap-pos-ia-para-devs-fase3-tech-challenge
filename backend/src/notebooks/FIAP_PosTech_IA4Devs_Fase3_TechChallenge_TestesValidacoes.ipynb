## Testes para validação da API

```python
import requests

API_URL = "https://shining-setting-trench.ngrok-free.dev/generate"

payload = {
    "pergunta": "Qual o protocolo para dor toracica?",
    "contexto": "",
    "max_new_tokens": 200
}

response = requests.post(API_URL, json=payload, timeout=60)
print("Status:", response.status_code)
print("Resposta:", response.json())
```

```python
# HEALTH CHECK
health = requests.get("https://shining-setting-trench.ngrok-free.dev/")
print(health.json())  # deve retornar {"status": "ok"}
```

```python
# TESTE DE RETORNO DE ERRO
resp_erro = requests.post(API_URL, json={"contexto": "sem pergunta"}, timeout=30)
print(resp_erro.status_code, resp_erro.json())
```
